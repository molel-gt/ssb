```python
import json
import os

import matplotlib.pyplot as plt
import numpy as np

import plot_opts
plt.rcParams.update(plot_opts.params)
```

```python
files_dir = "temp"
with open(os.path.join(files_dir, "i_x_density_0.2.json")) as f:
    t02 = json.load(f)
with open(os.path.join(files_dir, "i_x_density_0.4.json")) as f:
    t04 = json.load(f)
with open(os.path.join(files_dir, "i_x_density_0.6000000000000001.json")) as f:
    t06 = json.load(f)
with open(os.path.join(files_dir, "i_x_density_0.8.json")) as f:
    t08 = json.load(f)
with open(os.path.join(files_dir, "i_x_density_1.0.json")) as f:
    t10 = json.load(f)
```

```python
fig, ax = plt.subplots()
ax.plot(t02["i_min [A/m2]"], t02["areal density"], 'r-', label=t02["t [s]"])
# ax.plot(t04["i_min [A/m2]"], t04["areal density"], 'b-', label=t04["t [s]"])
ax.plot(t06["i_min [A/m2]"], t06["areal density"], 'g-', label=f"{t06["t [s]"]:.1f}")
# ax.plot(t08["i_min [A/m2]"], t08["areal density"], 'k-', label=f"{t08["t [s]"]:.1f}")
ax.plot(t10["i_min [A/m2]"], t10["areal density"], 'y-', label=f"{t10["t [s]"]:.1f}")
ax.set_box_aspect(1.0)
ax.legend(title="time [s]")
ax.set_xlabel(r"i [A/m$^2$]")
ax.set_ylabel("areal density")
ax.set_xlim([0, 50])
plt.tight_layout()
```
